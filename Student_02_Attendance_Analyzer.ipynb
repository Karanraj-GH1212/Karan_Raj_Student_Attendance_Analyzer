## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
df["Attendance_Percentage"] = (df["Classes_Attended"] / df["Classes_Held"]) * 100
df["Classes_Missed"] = df["Classes_Held"] - df["Classes_Attended"]
display(df)

# TODO: Filter records using a condition
below_75 = df[df["Attendance_Percentage"] < 75]

print("Students with attendance below 75%:")
display(below_75)

# TODO: Sort records and display the top/bottom results
sorted_data = df.sort_values(by="Attendance_Percentage", ascending=False)

#top5
print("Top 5 students:")
display(sorted_data.head())

#bottom5
print("Bottom 5 students:")
display(sorted_data.tail())
```

## 6. Your individual project challenge
**Challenge:** Calculate attendance percentage; identify students below 75%; find highest and lowest attendance.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Calculate attendance percentage
df["Attendance_Percentage"] = (df["Classes_Attended"] / df["Classes_Held"]) * 100

# Find students below 75% attendance
below_75 = df[df["Attendance_Percentage"] < 75]

print("Students below 75% attendance:")
display(below_75)

# Find highest attendance
highest = df.loc[df["Attendance_Percentage"].idxmax()]

print("Highest Attendance:")
display(highest)

# Find lowest attendance
lowest = df.loc[df["Attendance_Percentage"].idxmin()]

print("Lowest Attendance:")
display(lowest)


#explaination-->In this project... I calculated the attendance percentage of every student.
#I used filtering to find students whose attendance is below 75%.
#I also found the students with the highest and lowest attendance.
```
